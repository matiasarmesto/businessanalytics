## Create a Loop to Submit Reviews

➡️ Assignment Tasks
- Create a FOR loop to iterate through reviews
- Collect the responses in a single datafram

```python
#loop and collect responses
responses = []
for review in df['Review']:
  prompt = "Summarize the sentiment and most important points in the following user review for a phone company named Megatelco. Format the output in a table. "
  prompt = prompt + review
  response = model.generate_content(prompt)
  responses.append(response.candidates[0].content.parts[0].text)
  time.sleep(2) #pause so it doesnt overload
#new dataframe
response_df = pd.DataFrame({'Review Response': responses})
show(response_df)
```

## Summarize the Responses

➡️ Assignment Tasks
- Create at least two visualizations to show the most important themes and sentinment of the reviews

```python
#themes
import matplotlib.pyplot as plt
from collections import Counter

# Sample filler words
filler_words = ["the", "a", "an", "is", "are", "was", "were", "in", "on", "at", "to", "of", "and", "or", "but", "for", "with", "this", "that", "it", "its", "they", "them", "their", "we", "us", "our", "as", "by", "be", "have", "has", "had", "do", "does", "did", "can", "could", "will", "would", "should", "i", "you", "he", "she", "it", "me", "my", "your", "his", "her",
                'user','points','key','summarizing','points','sentiment','customer','table', 'overall','megatelco','summary','aspect','indicating','from']

keywords = []
for response in responses:
  words = response.lower().split()
  keywords.extend([word for word in words if word.isalpha() and len(word) > 2 and word not in filler_words]) #only words longer than 2 letters and not in filler words list

keyword_counts = Counter(keywords)
most_common_keywords = keyword_counts.most_common(10) #10 keywords

#barchart
keyword_labels, keyword_values = zip(*most_common_keywords)
plt.figure(figsize=(12, 6))
plt.bar(keyword_labels, keyword_values, color='skyblue')
plt.xlabel("Keywords")
plt.ylabel("Frequency")
plt.title("Most Frequent Keywords in Customer Review Responses (excluding filler words)")
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for readability
plt.tight_layout()  # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
#sentiment frequency
def analyze_sentiment(text):
    positive_keywords = ["excellent", "impressed", "great", "thrilled", "fantastic", "top-notch", "commendable", "satisfied"]
    negative_keywords = ["disappointed", "issues", "regrettably", "malfunctioning", "unresponsive", "constant", "problem", "glitch"]

    positive_score = sum(1 for keyword in positive_keywords if keyword in text.lower())
    negative_score = sum(1 for keyword in negative_keywords if keyword in text.lower())

    if positive_score > negative_score:
        return "Positive"
    elif negative_score > positive_score:
        return "Negative"
    else:
        return "Neutral"

sentiments = [analyze_sentiment(response) for response in response_df['Review Response']]
sentiment_counts = Counter(sentiments)
sentiment_labels, sentiment_values = zip(*sentiment_counts.items())

plt.figure(figsize=(8, 6))
plt.pie(sentiment_values, labels=sentiment_labels, autopct='%1.1f%%', startangle=90)
plt.title("Sentiment Analysis of Customer Reviews")
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
