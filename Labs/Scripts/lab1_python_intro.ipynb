## Getting data into Pandas

In this case we will load data from the statsmodels.org library

MTCARS is a traditional dataset used for learning.



```python
mtcars = sm.datasets.get_rdataset("mtcars", "datasets", cache=True).data
df = pd.DataFrame(mtcars)
df['carmodel'] = df.index
print(df)
```

## Your Turn: Create an Additional Chart

➡️ Assignment Tasks
- Create a pie chart showing the count of cars by transmission type
- Create a bar chart with different colors for each car model

```python
#Your code goes here

# Custom labels for 'am'
custom_labels = {0: 'Automatic Transmission', 1: 'Manual Transmission'}

# Group by 'am' and count 'carmodel'
car_counts = df.groupby('am')['carmodel'].count()

# Create the pie chart
car_counts.plot(
    kind='pie',
    labels=[custom_labels[am] for am in car_counts.index],  # Apply custom labels
    autopct=lambda p: f"{int(p * sum(car_counts) / 100)}",  # Show count instead of percentage
    title="Transmission Type",
    ylabel='',
    figsize=(6, 6)
)

plt.show()
```

```python
# Assuming 'df' is your DataFrame containing car model data
car_model_counts = df['carmodel'].value_counts()

# Generate 32 distinct colors using a colormap
colors = plt.cm.get_cmap('viridis', len(car_model_counts))
color_list = [colors(i) for i in range(len(car_model_counts))]

# Create the bar chart
plt.figure(figsize=(12, 6))  # Adjust figure size as needed
plt.bar(car_model_counts.index, car_model_counts.values, color=color_list)

plt.title('Cars by Model')
plt.xlabel('Car Model')
plt.ylabel('Count')
plt.xticks(rotation=90)  # Rotate x-axis labels for readability

plt.tight_layout()  # Adjust layout to prevent labels from overlapping
plt.show()
```
